```python
import pandas as pd, matplotlib.pyplot as plt, re

A1 = ["Allcott et al. (2020)","Hunt et al. (2018)","Pieh et al. (2025)","Pedersen et al. (2022)","Precht et al. (2024)",
      "L. Li et al. (2025)","Olson et al. (2023)","Holte and Ferraro (2023)","Kasturiratna et al. (2025)","Castelo et al. (2025)",
      "Zimmermann & Sobolev (2023)","Rahmillah et al. (2023)","Žmavc et al. (2025)","Rico-González et al. (2025)","Y. Wu et al. (2025)",
      "Lai et al. (2025)","Poonia et al. (2024)","Kaur et al. (2024)","Sanders et al. (2018)","Schmidt-Persson et al. (2024)",
      "Ho et al. (2025)","Z. Li et al. (2023)","Jones et al. (2021)","Conroy et al. (2017)","Lan et al. (2018)","Fendel et al. (2024)",
      "Lee et al. (2022)","Zhang et al. (2025)","Aldbyani et al. (2025)","R. Wu et al. (2024)",
      "Monge Roffarello and De Russis (2023)","Orzikulova et al. (2024)"]
A2 = ["Burnell et al. (2025)","Loid et al. (2020)","Dekker et al. (2025)","Oulasvirta et al. (2012)","Zimmermann (2021)",
      "Hiniker et al. (2016)","Vu and Tagliabue (2025)","Coyne and Woodruff (2023)","van Wezel et al. (2021)","Babic et al. (2016)",
      "Kaye et al. (2020)","May et al. (2025)","Radtke et al. (2022)"]
A3 = ["Rozgonjuk et al. (2018)","Hitcham et al. (2023)","Ellis et al. (2019)","Giansanti (2025)","Molaib et al. (2025)",
      "Parry et al. (2021)","Lu et al. (2025)","Ratan et al. (2025)","Poli et al. (2023)","O. Perez et al. (2023)"]

groups = [(A1,"What works (Table A1)"),(A2,"What fails (Table A2)"),(A3,"What's missing (Table A3)")]
df = pd.DataFrame([(p, int(re.search(r"\((\d{4})\)", p).group(1)), g) for L,g in groups for p in L],
                  columns=["paper","year","group"])
print("Total:", len(df))  # 55

years = list(range(df.year.min(), df.year.max()+1))
per = df.groupby("year").size().reindex(years, fill_value=0)
cum = per.cumsum()


fig, ax1 = plt.subplots(figsize=(10,5.5))
ax1.plot(years, per.values, "o--", color="#4472C4", ms=8, label="Papers published per year")
ax1.set_ylabel("Papers published per year", color="#4472C4"); ax1.tick_params(axis="y", labelcolor="#4472C4")
ax1.set_xticks(years); ax1.set_xlabel("Publication year"); ax1.grid(axis="y", alpha=0.3)
ax2 = ax1.twinx()
ax2.plot(years, cum.values, "o:", color="#70AD47", ms=8, label="Cumulative included papers")
ax2.set_ylabel("Cumulative number of included papers", color="#70AD47"); ax2.tick_params(axis="y", labelcolor="#70AD47")
ax2.set_ylim(0, 60)
fig.legend(loc="upper left", bbox_to_anchor=(0.09,0.9), frameon=False)
plt.title("Included papers by publication year (n = 55)"); plt.tight_layout()
plt.savefig("papers_by_year_n55.png", dpi=300); plt.show()

```
