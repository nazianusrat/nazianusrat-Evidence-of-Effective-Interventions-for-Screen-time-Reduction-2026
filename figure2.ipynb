```python
import matplotlib.pyplot as plt


types = ["Time limits &\nrestriction",
         "Device settings\n& design nudges",
         "Awareness, tracking\n& notifications",
         "Family / parental\nprograms",
         "School-based\nprograms",
         "Mindfulness, CBT\n& lifestyle",
         "AI-based /\npersonalized apps"]

improved  = [3, 3, 1, 4, 0, 3, 3]   # significant improvement
mixed     = [3, 1, 1, 1, 0, 0, 0]   # mixed results
no_effect = [1, 1, 3, 0, 1, 0, 0]   # no significant effect


fig, ax = plt.subplots(figsize=(10, 5.5))
ax.barh(types, improved, color="#2E7D32", label="Significant improvement")
ax.barh(types, mixed, left=improved, color="#F9A825",
        label="Mixed (faded, side effects or partial)")
start_red = [a + b for a, b in zip(improved, mixed)]
ax.barh(types, no_effect, left=start_red, color="#C62828",
        label="No significant effect")


for i in range(len(types)):
    start = 0
    for value in [improved[i], mixed[i], no_effect[i]]:
        if value > 0:
            ax.text(start + value / 2, i, str(value), ha="center",
                    va="center", color="white", fontweight="bold")
        start = start + value
    ax.text(start + 0.1, i, "k = " + str(start), va="center")


ax.invert_yaxis()
ax.set_xlim(0, 8)
ax.set_xlabel("Number of intervention study reports")
ax.xaxis.grid(True, color="lightgray")
ax.set_axisbelow(True)
ax.legend(loc="lower right", frameon=False)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.tight_layout()


plt.savefig("figure2.png", dpi=300)
plt.show()

from google.colab import files
files.download("figure2.png")
```
